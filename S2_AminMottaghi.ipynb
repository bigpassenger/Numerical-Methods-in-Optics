# 16.

Create two row vectors:  
a = 2 : 3 : 17  
b = 3 : 4 : 15   

Then, by only using the name of the vectors (a and b), create a row vector \( c \) that is made from the elements of \( a \) followed by the elements of \( b \).

```python
import numpy as np

a = np.arange(2, 17, 3)
b = np.arange(3, 16, 4)
print("a and b:\n", a,"\n" ,b)
print()
c = np.concatenate((a, b)) # Create c
print("c is equal to:\n", c)
```

# 21.

Create the following matrix by using vector notation for creating vectors with constant spacing and/or the `linspace` command. Do not type individual elements explicitly.

$$ A = 
\begin{bmatrix}
130 & 110 & 90 & 70 & 50 & 30 & 10 \\
1 & 2.8333 & 4.6667 & 6.5 & 8.3333 & 10.1667 & 12 \\
12 & 22 & 32 & 42 & 52 & 62 & 72
\end{bmatrix} $$
