# 6. Electric Field Intensity Along the Axis of a Charged Ring

The electric field intensity, \( E(z) \), due to a ring of radius \( R \) at any point \( z \) along the axis of the ring is given by:

$$
E(z) = \frac{\lambda}{2\epsilon_0} \cdot \frac{1}{(z^2 + R^2)^{3/2}}
$$

where  
- lambda is the charge density (C/m),  
-  epsilon_0 = 8.85 times 10^{-12} , C^{N·m}^2  is the electric constant,  
-  R  is the radius of the ring (m).

Consider the case where  
$$
\lambda = 1.7 \times 10^{-7} \, \text{C/m}, \qquad R = 6 \, \text{cm} = 0.06 \, \text{m}.
$$

### (a)
Determine  E(z)  at  z = 0, 2, 4, 6, 8, 10 , cm  (convert to meters when evaluating).

### (b)
Determine the distance  z  where E  is maximum.  
Do this by creating a vector  z  with elements ranging from  2cm  to  6cm  and spacing of  0.01cm.  
Calculate  E  for each value of  z  and then find the maximum  E  and associated  z  using MATLAB’s built‑in function `max`.

```python
import numpy as np

l = 1.7 * 10**(-7)        
R = 6 * 10**(-2)            
e = 8.85 * 10**(-12)        

def Electrical_field(lis):
    for z in lis:
        yield (l / (2 * e)) * (1 / ((z**2 + R**2) ** (3/2)))

# Part (a)
z_cm = [0, 2, 4, 6, 8, 10]
z_m = [x * 0.01 for x in z_cm]      
for y in Electrical_field(z_m):
    print(y)

# Part (b)
z_cm_b = np.arange(2.0, 6.01, 0.01)  
z_m_b = z_cm_b * 0.01                

maximum = -1e9
max_idx = 0
for idx, y in enumerate(Electrical_field(z_m_b)):
    if y > maximum:
        maximum = y
        max_idx = idx

print()
print(f"The maximum field is at {z_cm_b[idx]}m, equals to {maximum}")
```

# 16.

The area of a triangle $ABC$ can be calculated by

$$
\text{Area} = \frac{|\vec{r}_{AB} \times \vec{r}_{AC}|}{2}
$$

where $\vec{r}_{AB}$ and $\vec{r}_{AC}$ are vectors connecting the vertices $A$ and $B$, and $A$ and $C$, respectively. Determine the area of the triangle shown in the figure. Use the following steps in a script file to calculate the area. First, define the vectors $\vec{r}_{OA}$, $\vec{r}_{OB}$ and $\vec{r}_{OC}$ from knowing the coordinates of points $A$, $B$, and $C$. Then determine the vectors $\vec{r}_{AB}$ and $\vec{r}_{AC}$ from $\vec{r}_{OA}$, $\vec{r}_{OB}$ and $\vec{r}_{OC}$. Finally, determine the area by using MATLAB's built-in functions `cross`, `sum`, and `sqrt`.

```python
import numpy as np

# points
A = [8, 5, -4]
B = [-7, 9, 6]
C = [-5, -2, 11]

# points to vectors
ra = np.array(A)
rb = np.array(B)
rc = np.array(C)

# rab, rac
rab = ra - rb
rac = ra - rc

# area
cross_prod = np.cross(rab, rac)
abs = np.sqrt(np.sum(cross_prod ** 2))
area = abs / 2

print("Area of the triangle ABC is equal to: ", area)
```

# 21.

The position as a function of time \( (x(t), y(t)) \) of a projectile fired with a speed of \( v_0 \) at an angle \( \alpha \) is given by

$$
x(t) = v_0 \cos \alpha \cdot t \qquad y(t) = v_0 \sin \alpha \cdot t - \frac{1}{2} g t^2
$$

where \( g = 9.81 \, \text{m/s}^2 \). The polar coordinates of the projectile at time \( t \) are \( (r(t), \theta(t)) \), where

$$
r(t) = \sqrt{x(t)^2 + y(t)^2} \qquad \text{and} \qquad \tan \theta(t) = \frac{y(t)}{x(t)}.
$$

Consider the case where

$$
v_0 = 162 \, \text{m/s} \quad \text{and} \quad \alpha = 70^\circ.
$$

Determine r(t) and theta for \( t = 1, 6, 11, ...., 31 (s). \)

```python
from math import cos, sin, pi
import numpy as np

a = 70 * pi/180
v0 = 162
g = 9.81

def calculation(t):
    X_t = v0 * cos(a) * t
    Y_t = v0 * sin(a) * t - 0.5 * g * t**2
    tant = np.arctan2(Y_t, X_t)
    r_t = np.sqrt(X_t**2 + Y_t**2)
    return tant, r_t

t_ = np.arange(1, 32, 5)
for t in t_:
    theta, r = calculation(t)  
    print(f"t={t:2d}s, theta={np.degrees(theta):.2f}°, r={r:.2f}m")
```
