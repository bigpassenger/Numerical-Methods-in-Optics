###  Example 1 : Superposition of two plane waves

```python
from math import pi, tan
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

```python
N = 512
x0 = np.linspace(-1, 1, N) * 1e-3
y0 = np.linspace(-1, 1, N) * 1e-3

X, Y = np.meshgrid(x0,y0) # creating a mesh grid
lamb = 632e-9 # Wavelenght
k = 2*pi/ lamb  # Wavenumber
theta = 0.2 # angel of k-vectors with respect to z-axis
z = 0.05 # position of overlap
```

```python
a1 = 1 # amplitude of the first wave 
u1 = a1 * np.exp(-1j * k * (np.cos(np.radians(theta)) * z + np.sin(np.radians(theta)) * Y))
```

```python
a2 = 1 # amplitude of the second wave 
u2 = a2 * np.exp(-1j * k * (np.cos(np.radians(theta)) * z - np.sin(np.radians(theta)) * Y))
```

```python
u = u1 + u2 # superposition of the waves
I = abs(u) ** 2
```

```python
# plotting 
plt.figure(figsize=(8, 6))
plt.imshow(I, extent=[X.min()*1e3, X.max()*1e3, Y.min()*1e3, Y.max()*1e3],
           origin='lower', cmap='hot')
plt.colorbar(label='Intensity (a.u.)')
plt.xlabel('x (cm)')
plt.ylabel('y (cm)')
plt.title(f'Interference of two plane waves')

# visibility 
p = I[:, N // 2]                     
V = (np.max(p) - np.min(p)) / (np.max(p) + np.min(p))


# plot
plt.figure()
plt.plot(y0 * 1000, p, linewidth=2)   
plt.grid(True)
# plt.ylim(0, 4)
plt.xlabel('y (cm)')                  
plt.ylabel('Intensity')
plt.title(f'visibility')


plt.tight_layout()
plt.show()
```

### Example 6 : Interference of two LG beams - Collinear Propagation

```python
N = 512
x0 = np.linspace(-1, 1, N) *1e-2
y0 = np.linspace(-1, 1, N) *1e-2
la = 632.8e-9
k = 2*np.pi/la
X, Y = np.meshgrid(x0, y0)

xc = 0e-2
yc = 0e-2

def car2pol(x0, y0):
    r = np.sqrt(x0 ** 2 + y0 ** 2)
    th = np.arctan(y0 /x0)
    return r, th

r, th = car2pol(X - xc, Y - yc)
phi0 = 0 # a phase between u1, u2
w0 = 0.25e-2
GB = np.exp(-2 * r / w0)
```

```python
########################### First LG Wave  ###########################
def LG1(l, phi0):
    ax1 = 0.0e-3
    ay1 = 0.0
    a1 = 1
    u1 = (a1 * np.exp(-1j * k * (ax1*X + ay1*Y)) *
          (np.sqrt(2) * (r / w0)) ** (abs(l)) *
          np.exp(-1j * l * theta) * GB *
          np.exp(-1j * phi0))
    return u1
```

```python
def LG2(l):
    a2 = 1
    u2 = (a2 * (np.sqrt(2) * (r / w0)) ** (abs(l)) *
          np.exp(-1j * l * theta) * GB)
    return u2
```

```python
items = [(2, -3, 0), (-3, 2, 0), (2, -2, 0), (2, -2, pi), 
         (2, -3, -0.5*pi), (-3, 2, -0.5*pi), (1, -2, 0), (1, -2, 1.5*pi)]

fig, axes = plt.subplots(2, 4, figsize=(12, 6))  # width=12, height=6

# Adjust spacing between subplots
plt.subplots_adjust(left=0.08, right=0.92, top=0.88, bottom=0.08, 
                    wspace=0.3, hspace=0.4)


for ax, (l1, l2, phi0) in zip(axes.flat, items):
    u = LG1(l1, phi0) + LG2(l2)          # superposition 
    I = np.abs(u) ** 2
    
    # Plot 
    im = ax.imshow(I, extent=[x0[0]*1e3, x0[-1]*1e3, y0[0]*1e3, y0[-1]*1e3], origin='lower', 
                   cmap='hot', aspect='equal')
    
    if phi0 == 0:
        ax.set_title(f'ℓ₁={l1}, ℓ₂={l2}', fontsize=10)
    else:
        ax.set_title(f'ℓ₁={l1}, ℓ₂={l2}, φ₀={phi0/pi:.1f}π', fontsize=9)
    ax.set_xlabel('x')
    ax.set_ylabel('y')

# Add a colorbar
cbar = plt.colorbar(im, ax=axes, orientation='vertical', 
                    fraction=0.02, pad=0.02)
cbar.set_label('Intensity')

plt.suptitle('Superposition of LG modes', fontsize=14, y=0.98)
plt.show()
```
