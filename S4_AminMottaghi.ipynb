###  Example 1 : Superposition of two plane waves

```python
from math import pi, tan
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

```python
N = 512
x0 = np.linspace(-1, 1, N) * 1e-3
y0 = np.linspace(-1, 1, N) * 1e-3

X, Y = np.meshgrid(x0,y0) # creating a mesh grid
lamb = 632e-9 # Wavelenght
k = 2*pi/ lamb  # Wavenumber
theta = 0.2 # angel of k-vectors with respect to z-axis
z = 0.05 # position of overlap
```

```python
a1 = 1 # amplitude of the first wave 
u1 = a1 * np.exp(-1j * k * (np.cos(np.radians(theta)) * z + np.sin(np.radians(theta)) * Y))
```

```python
a2 = 1 # amplitude of the second wave 
u2 = a2 * np.exp(-1j * k * (np.cos(np.radians(theta)) * z - np.sin(np.radians(theta)) * Y))
```

```python
u = u1 + u2 # superposition of the waves
I = abs(u) ** 2
```

```python
# plotting 
plt.figure(figsize=(8, 6))
plt.imshow(I, extent=[X.min()*1e3, X.max()*1e3, Y.min()*1e3, Y.max()*1e3],
           origin='lower', cmap='hot')
plt.colorbar(label='Intensity (a.u.)')
plt.xlabel('x (cm)')
plt.ylabel('y (cm)')
plt.title(f'Interference of two plane waves')

# visibility 
p = I[:, N // 2]                     
V = (np.max(p) - np.min(p)) / (np.max(p) + np.min(p))


# plot
plt.figure()
plt.plot(y0 * 1000, p, linewidth=2)   
plt.grid(True)
# plt.ylim(0, 4)
plt.xlabel('y (cm)')                  
plt.ylabel('Intensity')
plt.title(f'visibility')


plt.tight_layout()
plt.show()
```

###  Example 4 : Superposition of two plane waves (in 3D space with respect into X-Y-Z angels)

```python
N = 512
x0 = np.linspace(-1, 1, N) *1e-3
y0 = np.linspace(-1, 1, N) * 1e-3

X, Y = np.meshgrid(x0, y0) # creating a mesh grid
lamb = 632e-9 # Wavelenght
k = 2*pi/ lamb  # Wavenumber
z = 0.05 # position of overlap
```

```python
a1 = 1 # amplitude of the first wave 
beta1 = 1 # angel of k-vectors with respect to x-axis
gama1 = 1.5 # angel of k-vectors with respect to y-axis
alpha1 = 2 # angel of k-vectors with respect to z-axis
u1 = a1 * np.exp(-1j * k * (np.cos(np.radians(beta1)) * X + np.cos(np.radians(gama1)) * Y + np.cos(np.radians(alpha1)) * z))
```

```python
a2 = 1 # amplitude of the first wave 
beta2 = 5 # angel of k-vectors with respect to x-axis
gama2 = 4 # angel of k-vectors with respect to y-axis
alpha2 = 4 # angel of k-vectors with respect to z-axis
u2 = a2 * np.exp(-1j * k * (np.cos(np.radians(beta2)) * X + np.cos(np.radians(gama2)) * Y + np.cos(np.radians(alpha2)) * z))
```

```python
u = u1 + u2 # superposition of the waves
I = abs(u) ** 2
```

```python
# plotting 
plt.figure(figsize=(8, 6))
plt.imshow(I,extent=[X.min()*1e3, X.max()*1e3, Y.min()*1e3, Y.max()*1e3],
           origin='lower', cmap='hot')
plt.colorbar(label='Intensity (a.u.)')
plt.xlabel('x (cm)')
plt.ylabel('y (cm)')
plt.title(f'Interference of two plane waves')


plt.tight_layout()
plt.show()
```
