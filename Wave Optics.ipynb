###  Example 1 : Superposition of two plane waves

```python
from math import pi, tan
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

```python
N = 512
x0 = np.linspace(-1, 1, N) * 1e-3
y0 = np.linspace(-1, 1, N) * 1e-3

X, Y = np.meshgrid(x0,y0) # creating a mesh grid
lamb = 632e-9 # Wavelenght
k = 2*pi/ lamb  # Wavenumber
theta = 0.2 # angel of k-vectors with respect to z-axis
z = 0.05 # position of overlap
```

```python
a1 = 1 # amplitude of the first wave 
u1 = a1 * np.exp(-1j * k * (np.cos(np.radians(theta)) * z + np.sin(np.radians(theta)) * Y))
```

```python
a2 = 1 # amplitude of the second wave 
u2 = a2 * np.exp(-1j * k * (np.cos(np.radians(theta)) * z - np.sin(np.radians(theta)) * Y))
```

```python
u = u1 + u2 # superposition of the waves
I = abs(u) ** 2
```

```python
# plotting 
plt.figure(figsize=(8, 6))
plt.imshow(I, extent=[X.min()*1e3, X.max()*1e3, Y.min()*1e3, Y.max()*1e3],
           origin='lower', cmap='hot')
plt.colorbar(label='Intensity (a.u.)')
plt.xlabel('x (cm)')
plt.ylabel('y (cm)')
plt.title(f'Interference of two plane waves')

# visibility 
p = I[:, N // 2]                     
V = (np.max(p) - np.min(p)) / (np.max(p) + np.min(p))


# plot
plt.figure()
plt.plot(y0 * 1000, p, linewidth=2)   
plt.grid(True)
# plt.ylim(0, 4)
plt.xlabel('y (cm)')                  
plt.ylabel('Intensity')
plt.title(f'visibility')


plt.tight_layout()
plt.show()
```

### Example 6 : Interference of two LG beams - Collinear Propagation

```python
N = 512
x0 = np.linspace(-1, 1, N) *1e-2
y0 = np.linspace(-1, 1, N) *1e-2
la = 632e-9
k = 2*np.pi/la
X, Y = np.meshgrid(x0, y0)

xc = 0e-2
yc = 0e-2

def car2pol(x0, y0):
    r = np.sqrt(x0 ** 2 + y0 ** 2)
    th = np.arctan2(y0, x0)
    return r, th

r, th = car2pol(X - xc, Y - yc)
# phi0 = 0 # a phase between u1, u2
w0 = 0.25e-2
GB = np.exp(-r**2 / w0**2)
```

```python
########################### First LG Wave  ###########################
def LG1(l, phi0):
    ax1 = 0.0
    ay1 = 0.0
    a1 = 1
    u1 = (a1 * np.exp(-1j * k * (ax1*X + ay1*Y)) *
          (np.sqrt(2) * (r / w0)) ** (abs(l)) *
          np.exp(-1j * l * th) * GB *
          np.exp(-1j * phi0))
    return u1
```

```python
def LG2(l):
    a2 = 1
    u2 = (a2 * (np.sqrt(2) * (r / w0)) ** (abs(l)) *
          np.exp(-1j * l * th) * GB)
    return u2
```

```python
items = [(2, -3, 0), (-3, 2, 0), (2, -2, 0), (2, -2, np.pi), 
         (2, -3, -0.5*np.pi), (-3, 2, -0.5*np.pi), (1, -2, 0), (1, -2, 1.5*np.pi)]

fig, axes = plt.subplots(2, 4, figsize=(12, 6))  # width=12, height=6

# Adjust spacing between subplots
plt.subplots_adjust(left=0.08, right=0.92, top=0.88, bottom=0.08, 
                    wspace=0.3, hspace=0.4)


for ax, (l1, l2, phi0) in zip(axes.flat, items):
    u = LG1(l1, phi0) + LG2(l2)          # superposition 
    I = np.abs(u) ** 2
    
    # Plot 
    im = ax.imshow(I, extent=[x0[0]*1e2, x0[-1]*1e2, y0[0]*1e2, y0[-1]*1e2], origin='lower', 
                   cmap='hot', aspect='equal')
    
    if phi0 == 0:
        ax.set_title(f'ℓ₁={l1}, ℓ₂={l2}', fontsize=10)
    else:
        ax.set_title(f'ℓ₁={l1}, ℓ₂={l2}, φ₀={phi0/pi:.1f}π', fontsize=9)
    ax.set_xlabel('x')
    ax.set_ylabel('y')

# Add a colorbar
cbar = plt.colorbar(im, ax=axes, orientation='vertical', 
                    fraction=0.02, pad=0.02)
cbar.set_label('Intensity')

plt.suptitle('Superposition of LG modes', fontsize=14, y=0.98)
plt.show()
```

### Example 7 : Temporal Coherence

```python
Nt = 128
N = 10000 
x0 = np.linspace(-1, 1, Nt) * 1e-2
y0 = np.linspace(-1, 1, Nt) * 1e-2
X, Y = np.meshgrid(x0, y0)

la = 632e-9
k = 2 * np.pi / la
ax1 = 5e-5
ax2 = -ax1
a1 = 1
a2 = 1

S = np.zeros_like(X)               
H = np.zeros((Nt, N))               
```

```python
for i in range(N):
    u1 = a1 * np.exp(-1J* (k * ax1 *X + (np.random.rand()-0.5)*2*pi))
    u2 = a2 * np.exp(-1J* (k * ax2 *X + (np.random.rand()-0.5)*2*pi))
    # u1 = a1 * np.exp(-1J* (k * ax1 *X + (0.5)*2*pi))
    # u2 = a2 * np.exp(-1J* (k * ax2 *X + (0.5)*2*pi))
    U = u1 + u2
    I = np.abs(U) ** 2
    H[:,i] = I[Nt//2, :]
    S = S + I
```

```python
It = S / Nt

# Plot H (central row evolution over realizations)
x_cm = x0 * 100
y_cm = y0 * 100
# Use extent to map pixel indices to real coordinates
extent = [x_cm[0], x_cm[-1], y_cm[0], y_cm[-1]]
plt.figure()
plt.imshow(H, extent=extent,aspect='auto', cmap='viridis')
plt.colorbar(label='Intensity')
plt.xlabel('Realization number')
plt.ylabel('x index')
plt.title('H: central row over time')
plt.show()

# Plot the final averaged intensity with physical coordinates
plt.figure()
# Convert coordinates from meters to cm for plotting
x_cm = x0 * 100
y_cm = y0 * 100
# Use extent to map pixel indices to real coordinates
extent = [x_cm[0], x_cm[-1], y_cm[0], y_cm[-1]]
plt.imshow(It, extent=extent, origin='lower', cmap='viridis', vmin=0, vmax=4)
plt.colorbar(label='Intensity')
plt.xlabel('x (cm)')
plt.ylabel('y (cm)')
plt.title('Time-averaged intensity')
plt.axis('image')            # equivalent to MATLAB's axis image
plt.xlim([-1, 1])
plt.ylim([-1, 1])
plt.tick_params(labelsize=16)
plt.show()
```

### Problem 1: Generate the fringe pattern of the interference of an LG beam (several l values) with a Bessel beam

```python
N = 512
x0 = np.linspace(-1, 1, N) *1e-2
y0 = np.linspace(-1, 1, N) *1e-2
la = 632e-9
k = 2*np.pi/la
X, Y = np.meshgrid(x0, y0)
alpha = 10000
phi0 = 0
xc = 0e-2
yc = 0e-2
# xc = 0
# yc = 0
def car2pol(x0, y0):
    r = np.sqrt(x0 ** 2 + y0 ** 2)
    th = np.arctan2(y0, x0)
    return r, th

r, th = car2pol(X - xc, Y - yc)
# phi0 = 0 # a phase between u1, u2
w0 = 0.25e-2
GB = np.exp(-r**2 / w0**2)
```

```python
########################### First LG Wave  ###########################
def LG1(l, phi0):
    ax1 = 0.0
    ay1 = 0.0
    a1 = 1
    u1 = (a1 * np.exp(-1j * k * (ax1*X + ay1*Y)) *
          (np.sqrt(2) * (r / w0)) ** (abs(l)) *
          np.exp(-1j * l * th) * GB *
          np.exp(-1j * phi0))
    return u1
```

```python
########################### Bessel beam using an axicon ###########################
def Bessel(alpha):
    w0 = 1e-3
    u2 = np.exp(-(r/w0)**2)
    axicon = np.exp(1J*alpha*r) 
    u2 = u2 * axicon
    return u2
```

```python
# Topological charges to loop over
items = [1, 2, 3, 4, 5, 6, 7, 8]

# Create 2x4 subplots
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
plt.subplots_adjust(left=0.08, right=0.92, top=0.88, bottom=0.08,
                    wspace=0.3, hspace=0.4)

# Flatten axes array for easy indexing
axes_flat = axes.flatten()

# Store the last image for the colorbar
last_im = None

for idx, l1 in enumerate(items):
    u = LG1(l1, phi0) + Bessel(alpha)   # superposition
    I = np.abs(u)**2
    
    # Plot on the corresponding subplot
    ax = axes_flat[idx]
    im = ax.imshow(I, extent=[x0[0]*1e2, x0[-1]*1e2, y0[0]*1e2, y0[-1]*1e2],
                   origin='lower', cmap='hot', aspect='equal')
    last_im = im   # keep for colorbar
    
    # Title – note: no l2, just l1
    ax.set_title(f'ℓ₁ = {l1}', fontsize=10)
    ax.set_xlabel('x (cm)')
    ax.set_ylabel('y (cm)')

# Add a single colorbar for the last image (works since all share same scale)
cbar = plt.colorbar(last_im, ax=axes, orientation='vertical',
                    fraction=0.02, pad=0.02)
cbar.set_label('Intensity')

plt.suptitle('Superposition of LG modes (ℓ₁) with a Bessel beam', fontsize=14, y=0.98)
plt.show()
```

### Problem 2 : Generate the fringe pattern of the interference of a paraboloidal wave with an off-centered LG beam (several l values).

```python
N = 512
x0 = np.linspace(-1, 1, N) * 1e-2
y0 = np.linspace(-1, 1, N) * 1e-2
X, Y = np.meshgrid(x0, y0)

la = 632e-9
k = 2*np.pi/la
phi0 = 0
xc = 0.0001
yc = 0.0001

def car2pol(x0, y0):
    r = np.sqrt(x0 ** 2 + y0 ** 2)
    th = np.arctan2(y0, x0)
    return r, th

r, th = car2pol(X - xc, Y - yc)
w0 = 0.25e-2
GB = np.exp(-r**2 / w0**2)
```

```python
#------------------------- first wave as polynomial wavefront --------------------------
def paraboloidal_w(X, Y):
    cx=[0, 1e5, 1e5, 0, 0]
    cy=[0, 1e6, 1e5, 0, 0]
    P=np.polyval(cx,X)+np.polyval(cy,Y)

    a2=1
    u1=a2*np.exp(-1j*P)
    return u1
```

```python
########################### Second wave as LG Wave  ###########################
def LG1(l, phi0):
    ax1 = 0.0
    ay1 = 0.0
    a1 = 1
    u2 = (a1 * np.exp(-1j * k * (ax1*X + ay1*Y)) *
          (np.sqrt(2) * (r / w0)) ** (abs(l)) *
          np.exp(-1j * l * th) * GB *
          np.exp(-1j * phi0))
    return u2
```

```python
# Topological charges to loop over
items = [1, 2, 3, 4, 5, 6, 7, 8]

# Create 2x4 subplots
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
plt.subplots_adjust(left=0.08, right=0.92, top=0.88, bottom=0.08,
                    wspace=0.3, hspace=0.4)

# Flatten axes array for easy indexing
axes_flat = axes.flatten()

# Store the last image for the colorbar
last_im = None
parabo = paraboloidal_w(X, Y) 
for idx, l1 in enumerate(items):
    u = LG1(l1, phi0) + parabo  # superposition
    I = np.abs(u)**2
    
    # Plot on the corresponding subplot
    ax = axes_flat[idx]
    im = ax.imshow(I, extent=[x0[0]*1e2, x0[-1]*1e2, y0[0]*1e2, y0[-1]*1e2],
                   origin='lower', cmap='hot', aspect='equal')
    last_im = im   # keep for colorbar
    
    # Title – note: no l2, just l1
    ax.set_title(f'ℓ₁ = {l1}', fontsize=10)
    ax.set_xlabel('x (cm)')
    ax.set_ylabel('y (cm)')

# Add a single colorbar for the last image (works since all share same scale)
cbar = plt.colorbar(last_im, ax=axes, orientation='vertical',
                    fraction=0.02, pad=0.02)
cbar.set_label('Intensity')

plt.suptitle('Superposition of LG modes (ℓ₁) with a paraboloidal  wave', fontsize=14, y=0.98)
plt.show()
```
