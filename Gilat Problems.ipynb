# 20.

The parametric equations of a line in space are:  
$$ x = x_0 + at, \quad y = y_0 + bt, \quad z = z_0 + ct $$

The distance \( d \) from a point \( A(x_A, y_A, z_A) \) to the line can be calculated by:

$$ d = d_{A0} \sin \left[ \arccos \left( \frac{(x_A - x_0)a + (y_A - y_0)b + (z_A - z_0)c}{d_{A0} \sqrt{a^2 + b^2 + c^2}} \right) \right] $$

where 
$$ d_{A0} = \sqrt{(x_A - x_0)^2 + (y_A - y_0)^2 + (z_A - z_0)^2} $$

Determine the distance of the point \( A(2, -3, 1) \) from the line 
$$ x = -4 + 0.6t, \quad y = -2 + 0.5t, \quad z = -3 + 0.7t $$

First define the variables \( x_0, y_0, z_0, a, b, c \), then use the variables (and the coordinates of point \( A \)) to calculate \( d_{A0} \), and finally calculate \( d \).

```python
# just want to use regex to be an extra exercise for myself 
import re
import math

# point
xA, yA, zA = (2, -3, 1)

# equations
eq_x = "x = -4 + 0.6t"
eq_y = "y = -2 + 0.5t"
eq_z = "z = -3 + 0.7t"

# finding x0, y0, z0 and a, b, c
def use_re(eq):
    pattern = r"=\s*([+-]?\d*\.?\d+)\s*\+\s*([+-]?\d*\.?\d+)\s*t"
    match = re.search(pattern, eq)
    if match:
        return float(match.group(1)), float(match.group(2))
    else:
        raise ValueError(f"Could not parse: {eq}")

# values
x0, a = use_re(eq_x)
y0, b = use_re(eq_y)
z0, c = use_re(eq_z)

# calculating d0 and d
def d0(x0, y0, z0, xA, yA, zA):
    dx = xA - x0
    dy = yA - y0
    dz = zA - z0
    return math.sqrt(dx*dx + dy*dy + dz*dz)
def d(x0, y0, z0, xA, yA, zA,d_A0, a, b, c):
    dx = xA - x0
    dy = yA - y0
    dz = zA - z0
    dir_norm = math.sqrt(a*a + b*b + c*c)
    ans = (dx*a + dy*b + dz*c) / (d_A0 * dir_norm)
    ans = d_A0 * math.sin(math.acos(ans))
    return ans


# distance
d_A0 = d0(x0, y0, z0, xA, yA, zA)
dis = d(x0, y0, z0, xA, yA, zA,d_A0, a, b, c)

print(dis)
```

# 25.

The voltage difference \( V_{ab} \) between points \( a \) and \( b \) in the Wheatstone bridge circuit is:

$$ V_{ab} = V \left( \frac{R_1 R_3 - R_2 R_4}{(R_1 + R_2)(R_3 + R_4)} \right) $$

Calculate the voltage difference when \( V = 14 \) volts,  
R_1 = 120.6,  R_2 = 119.3    
R_3 = 121.2 , and R_4 = 118.8 

```python
# params
R1 = 120.6
R2 = 119.3
R3 = 121.2
R4 = 118.8
V = 14

# again use function for Vab
def Voltage(R1, R2, R3, R4, V):
    V_ab = V * (((R1*R2) - (R2*R4)))/((R1+R2)*(R3 + R4))
    return V_ab

# calculation
ans = Voltage(R1, R2, R3, R4, V)
print(ans)
```

$$ V_C = V_0 (1 - e^{-t / (RC)}) $$

Given  Vc = 36 , V,  R = 2500 , and C = 1600 , calculate the current 8 seconds after the switch is closed.

# 35.

The value \( B \) of a principal \( P \) that is deposited in a saving account with a fixed annual interest rate \( r \) after \( n \) years can be calculated by the formula:

$$ B = P \left(1 + \frac{r}{m}\right)^{nm} $$

where \( m \) is the number of times that the interest is compounded annually. Consider a $80,000 deposit for 5 years. Determine how much more money will be earned if the interest rate is compounded monthly instead of annually.

```python
from math import exp 

# params
P = 80000       
r = 0.05        
t = 5           

def amount(P, r, m, t):
    return P * (1 + r / m) ** (m * t)

# amount
annual = amount(P, r, 1, t)

# monthly amount
monthly = amount(P, r, 12, t)

#earned 
difference = monthly - annual
print(difference)
```

# 40.

Stirling's approximation for large factorials is given by:

$$ n! = \sqrt{2\pi n} \left(\frac{n}{e}\right)^n $$

Use the formula for calculating \( 20! \). Compare the result with the true value obtained with MATLAB's built-in function `factorial` by calculating the error:

$$ \text{Error} = \frac{\text{TrueVal} - \text{ApproxVal}}{\text{TrueVal}} $$

```python
import math

#stirling
def stirling(n):
    return math.sqrt(2*math.pi*n)*(n /math.e) ** n

n = 20
approx_val = stirling(n)
true_val = math.factorial(n) #python's function
error = (true_val - approx_val) / true_val

print(error)
```
